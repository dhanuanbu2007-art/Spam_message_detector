```python
import nltk
import re
from collections import Counter
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
feedback = input("Enter your feedback: ")

print("\nYour feedback:")
print(feedback)
```

```python
text = feedback.lower()

print(text)
```

```python
text = re.sub(r'[^\w\s]', '', text)

print(text)
```

```python
words = text.split()

print(words)
```

```python
print("Total number of words:", len(words))
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)

print(filtered_words)
```

```python
word_frequency = Counter(filtered_words)

print(word_frequency)
```

```python
print("Most common words:")

for word, count in word_frequency.most_common(5):
    print(word, ":", count)
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()
```

```python
sentiment_scores = analyzer.polarity_scores(feedback)

print(sentiment_scores)
```

```python
score = sentiment_scores['compound']

if score >= 0.05:
    sentiment = "Positive 😊"
elif score <= -0.05:
    sentiment = "Negative 😞"
else:
    sentiment = "Neutral 😐"

print("Sentiment:", sentiment)
```

```python
import nltk

nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

words = word_tokenize(text)

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word.lower() not in stop_words and word.isalpha()
]

print("Original Text:")
print(text)

print("\nTokens:")
print(words)

print("\nAfter Stopword Removal:")
print(filtered_words)
```

```python
from collections import Counter

word_frequency = Counter(filtered_words)

print("Word Frequency:")
print(word_frequency)

print("\nMost Common Words:")
print(word_frequency.most_common(10))
```

```python
from textblob import TextBlob

analysis = TextBlob(text)

print("Sentiment Score:", analysis.sentiment.polarity)
print("Subjectivity:", analysis.sentiment.subjectivity)
```

```python
polarity = analysis.sentiment.polarity

if polarity > 0:
    sentiment = "Positive"
elif polarity < 0:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

print("Sentiment:", sentiment)
```
