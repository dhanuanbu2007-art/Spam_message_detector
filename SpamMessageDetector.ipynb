```python
import gradio as gr
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

# Sample training dataset
messages = [
    "Congratulations you have won a free prize",
    "Claim your free reward now",
    "You won a lottery",
    "Click here to claim your cash prize",
    "Win a free mobile phone today",
    "Urgent! Claim your reward immediately",
    "Get free offers and discounts now",
    "You have been selected for a cash award",
    "Congratulations! You are a lucky winner",
    "Free entry in a competition",
    "Your account has won a bonus",
    "Claim your exclusive offer today",
    "Limited time offer click now",
    "Earn money quickly from home",
    "You have received a free gift",
    "Get a loan instantly",
    "Call now to claim your prize",
    "You won a free vacation",
    "Special offer just for you",
    "Click this link to win money",

    "Hello how are you",
    "Are you coming to college tomorrow",
    "Please call me when you are free",
    "I will meet you at the bus stop",
    "Your assignment is due tomorrow",
    "Can you send me the notes",
    "Thank you for your help",
    "Let's meet this evening",
    "I am going to the library",
    "Please bring your notebook",
    "Happy birthday have a great day",
    "I will call you later",
    "Are we having class today",
    "Please share the project file",
    "See you tomorrow",
    "I have reached home safely",
    "Can we discuss the project",
    "The meeting starts at ten",
    "Please check your email",
    "Have a nice day"
]

labels = [
    "Spam"
] * 20 + [
    "Not Spam"
] * 20

# Create and train the machine learning model
model = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("classifier", MultinomialNB())
])

model.fit(messages, labels)


# Prediction function
def detect_spam(text):

    if not text.strip():
        return "Please enter a message."

    prediction = model.predict([text])[0]
    probabilities = model.predict_proba([text])[0]

    confidence = max(probabilities) * 100

    if prediction == "Spam":
        return f"""
## 🚨 Spam Message Detected!

**Classification:** Spam

**Confidence:** {confidence:.2f}%

This message may contain unwanted or promotional content.
Please verify the sender before clicking any links.
"""
    else:
        return f"""
## ✅ Normal Message

**Classification:** Not Spam

**Confidence:** {confidence:.2f}%

This message appears to be a normal message.
"""


# Gradio interface
demo = gr.Interface(
    fn=detect_spam,

    inputs=gr.Textbox(
        lines=5,
        label="Enter Your Message",
        placeholder="Type or paste a message here..."
    ),

    outputs=gr.Markdown(
        label="Detection Result"
    ),

    title="📩 Spam Message Detector",

    description="A Machine Learning application that classifies messages as Spam or Not Spam.",

    examples=[
        ["Congratulations! You won a free prize. Click here to claim now."],
        ["Hello, are you coming to college tomorrow?"],
        ["You have won a lottery. Claim your reward immediately."],
        ["Please send me the assignment notes."]
    ]
)

demo.launch(share=True)
```
